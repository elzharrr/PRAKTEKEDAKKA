**PROJECT CHARTER**

**1. DATA LOADING**

```python
df = pd.read_csv("data_kantin.csv")
```

**2. DATA INSPECTION**

```python
print("=== 5 BARIS PERTAMA ===")
print(df.head())

print("\n=== INFORMASI DATA ===")
print(df.info())

print("\n=== STATISTIK DATA ===")
print(df.describe())

print("\n=== UKURAN DATA ===")
print("Jumlah baris dan kolom:", df.shape)
```

**3. DATA CLEANING**

```python
print("\n=== MISSING VALUE SEBELUM CLEANING ===")
print(df.isnull().sum())

print("\nJumlah data duplikat:", df.duplicated().sum())


# Menghapus data duplikat
df = df.drop_duplicates()


# Mengisi missing value pada kolom terjual
# Menggunakan median agar tidak terlalu dipengaruhi
# oleh nilai penjualan yang terlalu tinggi/rendah
df["terjual"] = df["terjual"].fillna(df["terjual"].median())


# Mengisi missing value pada kolom menu
df["menu"] = df["menu"].fillna("Tidak Diketahui")


# Mengubah tipe data tanggal
df["tanggal"] = pd.to_datetime(df["tanggal"])


# Memastikan harga dan terjual berupa integer
df["harga"] = df["harga"].astype(int)
df["terjual"] = df["terjual"].astype(int)


# Mengecek hasil cleaning
print("\n=== MISSING VALUE SETELAH CLEANING ===")
print(df.isnull().sum())

print("\nJumlah duplikat setelah cleaning:", df.duplicated().sum())

print("\n=== TIPE DATA SETELAH CLEANING ===")
print(df.dtypes)
```

**4. DATA MANIPULATION**

```python
print("\n=== FILTER: PENJUALAN LEBIH DARI 20 ===")

df_filter = df[df["terjual"] > 20]

print(df_filter)

```

```python
print("\n=== SORT: PENJUALAN TERBANYAK ===")

df_sort = df.sort_values(
    by="terjual",
    ascending=False
)

print(df_sort.head(10))
```

```python
df["pendapatan"] = df["harga"] * df["terjual"]

print("\n=== KOLOM TURUNAN: PENDAPATAN ===")
print(df.head())
```

```python
print("\n=== RINGKASAN PENJUALAN PER MENU ===")

ringkasan_menu = df.groupby("menu").agg(
    total_terjual=("terjual", "sum"),
    total_pendapatan=("pendapatan", "sum")
).sort_values(
    by="total_terjual",
    ascending=False
)

print(ringkasan_menu)
```

**5. DATA PROFILING SUMMARY**

```python
print("\n=== DATA PROFILING SUMMARY ===")

menu_terlaris = ringkasan_menu["total_terjual"].idxmax()
jumlah_terjual = ringkasan_menu["total_terjual"].max()

menu_pendapatan = ringkasan_menu["total_pendapatan"].idxmax()
pendapatan_terbesar = ringkasan_menu["total_pendapatan"].max()

print(
    f"1. Menu dengan jumlah penjualan terbanyak adalah "
    f"{menu_terlaris} dengan total {jumlah_terjual} unit terjual."
)

print(
    f"2. Menu dengan pendapatan terbesar adalah "
    f"{menu_pendapatan} dengan total pendapatan "
    f"Rp {pendapatan_terbesar:,.0f}."
)

print(
    f"3. Setelah proses cleaning, dataset memiliki "
    f"{len(df)} baris dan {len(df.columns)} kolom."
)
```

**6. MENYIMPAN DATASET BERSIH**

```python
df.to_csv("dataset_bersih.csv", index=False)

print("\n=== SELESAI ===")
print("Dataset bersih berhasil disimpan sebagai 'dataset_bersih.csv'.")
```
